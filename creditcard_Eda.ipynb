# 📄 Dataset Source and Description

The dataset used in this project is the **Credit Card Fraud Detection** dataset published by the **Machine Learning Group (MLG), Université Libre de Bruxelles (ULB)** in collaboration with **Worldline**. It contains anonymized credit card transactions made by European cardholders during **September 2013**. To protect customer privacy, the original transaction features were transformed using **Principal Component Analysis (PCA)**, resulting in anonymized features **V1–V28**.

This dataset is widely used in machine learning research for **credit card fraud detection** because it represents a highly **imbalanced binary classification problem**, where fraudulent transactions account for only **492 out of 284,807 transactions (0.172%)**.

# 📚 Literature Review

## Overview

Credit card fraud has become one of the most significant challenges in the modern financial industry due to the rapid growth of online payments and digital banking. Fraudulent transactions can result in substantial financial losses for both customers and financial institutions. Traditional fraud detection methods, such as rule-based systems, are no longer sufficient because fraud patterns continuously evolve. Consequently, researchers have increasingly adopted machine learning and deep learning techniques to develop intelligent fraud detection systems.

The research paper **"Credit Card Fraud Detection" (Shah & Makwana, 2023)** reviews various fraud detection approaches and highlights the advantages of data-driven machine learning models over conventional rule-based systems.

---

# 🔍 Key Findings from the Literature

### 1. Growth of Credit Card Fraud

The increasing adoption of digital payment systems has significantly raised the risk of fraudulent transactions. Modern fraud techniques have become more sophisticated, making early detection essential for minimizing financial losses.

---

### 2. Limitations of Rule-Based Systems

Traditional fraud detection systems rely on predefined rules to identify suspicious transactions. Although these systems are simple to implement, they struggle to detect newly emerging fraud patterns because they cannot automatically adapt to changing attacker behavior.


### 3. Deep Learning Approaches

Recent studies have introduced deep learning models for fraud detection, including:

- Convolutional Neural Networks (CNN)
- Recurrent Neural Networks (RNN)
- Autoencoders
- Deep Neural Networks (DNN)

These models can automatically extract complex transaction patterns and improve fraud detection accuracy, particularly when large-scale transaction datasets are available.

---

### 4. Challenges in Fraud Detection

The literature identifies several major challenges:

- Highly imbalanced datasets
- Real-time fraud detection requirements
- Feature selection and engineering
- Model interpretability for regulatory compliance
- Computational cost of advanced machine learning models

Among these, **class imbalance** is considered the most critical challenge because fraudulent transactions represent only a very small percentage of total transactions.

---

### 5. Importance of Feature Engineering

Effective feature engineering improves model performance by selecting the most informative variables and removing irrelevant features. Proper preprocessing, normalization, and dimensionality reduction contribute significantly to better fraud detection.

### Principal Component Analysis (PCA)

**Principal Component Analysis (PCA)** is a statistical dimensionality reduction technique that transforms the original features into a new set of uncorrelated variables called **principal components**. It helps preserve the most important information in the data while protecting sensitive information and reducing complexity.

### Why were duplicate rows removed?

Duplicate records represent the **same transaction appearing more than once**. Keeping duplicate rows can:

- Bias statistical summaries.
- Distort feature distributions.
- Produce misleading visualizations.
- Overrepresent certain transactions.
- Reduce the reliability of EDA findings.

Removing duplicates ensures that **each transaction is counted only once**, resulting in more accurate and trustworthy analysis.

---

## ✅ 4. Outlier Treatment

No outliers were removed during the cleaning process.

**Reason:**

This dataset is designed for **credit card fraud detection**, where unusual or extreme transaction values may represent actual fraudulent behavior. Removing these observations could eliminate valuable fraud-related information.

Additionally, **V1–V28** are PCA-transformed features, so their extreme values are expected and should be preserved.

---

The columns **V1–V28** are generated using **Principal Component Analysis (PCA)**.

````python
# Convert Seconds into Hours
df['Hour'] = (df['Time'] // 3600).astype(int)

# Check
df[['Time','Hour']].head()
````

### Creating the Hour Feature

The original **Time** column records the number of seconds elapsed since the first transaction in the dataset, making it difficult to interpret transaction timing directly.

To improve readability and enable time-based analysis, a new feature **Hour** was created by converting seconds into hourly intervals using integer division. This feature is useful for analyzing transaction patterns and fraud occurrence across different hours of the day.

````python
# Style
sns.set_style("whitegrid")

# Create Hour Column
df['Hour'] = (df['Time'] // 3600).astype(int)

# Create Figure
fig, axes = plt.subplots(
    2, 1,
    figsize=(14,9),
    gridspec_kw={'height_ratios':[3,2]}
)

# =====================================
# Histogram of Transaction Time
# =====================================
sns.histplot(
    data=df,
    x='Time',
    bins=60,
    color='royalblue',
    edgecolor='black',
    ax=axes[0]
)

axes[0].set_title(
    'Distribution of Transaction Time',
    fontsize=18,
    fontweight='bold'
)

axes[0].set_xlabel('Time (Seconds from First Transaction)', fontsize=12)
axes[0].set_ylabel('Number of Transactions', fontsize=12)

# =====================================
# Fraud Transactions by Hour
# =====================================
ax = sns.countplot(
    data=df[df['Class']==1],
    x='Hour',
    color='crimson',
    edgecolor='black',
    ax=axes[1]
)

axes[1].set_title(
    'Fraud Transactions by Hour',
    fontsize=18,
    fontweight='bold'
)

axes[1].set_xlabel('Hour', fontsize=12)
axes[1].set_ylabel('Fraud Count', fontsize=12)

# Add Labels
for p in ax.patches:
    height = p.get_height()
    if height > 0:
        ax.annotate(
            f'{int(height)}',
            (p.get_x() + p.get_width()/2, height),
            ha='center',
            va='bottom',
            fontsize=9,
            fontweight='bold'
        )

# Dataset Information
fig.text(
    0.99,
    0.98,
    f"Total Transactions: {len(df):,} | Fraud Cases: {df['Class'].sum():,}",
    ha='right',
    fontsize=11,
    bbox=dict(facecolor='white', edgecolor='black')
)

plt.tight_layout()
plt.show()
````

### Findings

- Transaction activity is spread across the entire observation period, with noticeable peaks around **75,000–80,000 seconds** and **125,000–160,000 seconds**.
- Fraud transactions occurred throughout the **48-hour** period, with the highest fraud count of **43 transactions** recorded at **Hour 11**, followed by **27 transactions** at **Hour 27**.
- Most other hours recorded fewer than **20 fraud cases**, indicating that fraudulent transactions are relatively rare but occur consistently over time.
- The **Hour** feature provides a clearer view of temporal fraud patterns and helps identify periods with relatively higher fraud activity.

````python
# Style
sns.set_style("whitegrid")

# Count transactions by Hour and Class
hourly = (
    df.groupby(['Hour', 'Class'])
      .size()
      .reset_index(name='Count')
)

# Rename Class values
hourly['Class'] = hourly['Class'].replace({
    0: 'Normal',
    1: 'Fraud'
})

# Plot
plt.figure(figsize=(13,6))

sns.lineplot(
    data=hourly,
    x='Hour',
    y='Count',
    hue='Class',
    marker='o',
    linewidth=2.5
)

plt.title(
    'Fraud vs Normal Transactions by Hour',
    fontsize=18,
    fontweight='bold'
)

plt.xlabel('Hour of the Day', fontsize=12)
plt.ylabel('Number of Transactions', fontsize=12)

plt.xticks(range(24))
plt.grid(alpha=0.3)
plt.legend(title='Transaction Class')

plt.tight_layout()
plt.show()
````

### Findings

- Normal transactions are consistently higher than fraud transactions across all **48 hours**, with transaction counts peaking at around **9,800** near **Hour 20**.
- Fraud transactions remain very low throughout the observation period but show slight increases around **Hours 11, 27, and 42**.
- The clear difference between the two lines highlights the **severe class imbalance**, where fraudulent transactions represent only a tiny fraction of total transactions.
- Although fraud occurs across all hours, certain time periods show relatively higher fraud activity, making **Hour** a useful feature for fraud analysis.

````python
# Correlation with Target
corr_target = (
    df.corr(numeric_only=True)[['Class']]
      .sort_values(by='Class', ascending=False)
)

plt.figure(figsize=(6,10))

sns.heatmap(
    corr_target,
    annot=True,
    cmap='coolwarm',
    center=0,
    fmt=".2f",
    linewidths=0.5
)

plt.title(
    'Correlation of Features with Fraud (Class)',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
````

### Findings

- The strongest **positive correlations** with **Class** are **V11 (0.15)**, **V4 (0.13)**, and **V2 (0.08)**, indicating these features increase with fraudulent transactions.
- The strongest **negative correlations** are **V17 (-0.31)**, **V14 (-0.29)**, **V12 (-0.25)**, and **V10 (-0.21)**, making them the most informative features for fraud detection.
- Features such as **Time (-0.01)**, **Hour (-0.01)**, and **Amount (0.01)** have almost no linear relationship with the target variable.
- Overall, the heatmap suggests that a few PCA-transformed features carry most of the fraud-related information, while the majority of features have weak correlations with **Class**.
